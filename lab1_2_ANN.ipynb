```python
print("Task 1 ============================================================================")
#a) Create three variables to store your name, roll number, and CGPA. Print their values along with their data types using type().
#b) Create two integer variables marks1 and marks2. Use the += and -= operators to increase marks1 by 5 and decrease marks2 by 3, then print both updated values.
#c) Compare marks1 and marks2 using both the == and > operators, and print the results.
#d) Swap the values of marks1 and marks2 without using a third (temporary) variable, and print the values before and after swapping.


name = "Shumaim Iftikhar"
roll_no = "FA23-BAI-048"
cgpa = 3.32
print(name, type(name))
print(roll_no, type(roll_no))
print(cgpa, type(cgpa))


mark1=10
mark2=20
mark1 +=5
mark2-=3
print(mark1, mark2)

if mark1 >= mark2:
  print("true")
else:
  print("false")


  temp=mark1
  mark1=mark2
  mark2=temp
  print(mark1, mark2)


print("Task 2 ============================================================================")
#a) Create a list named favorite_subjects containing the names of five subjects you like.
#b) Add two new subjects to the list using append() and insert().
#c) Remove one subject from the list using remove(), and one using pop().
#d) Sort the final list alphabetically and print it, along with the total number of subjects using len().

favorite_subjects =["ENGLISH", "ARTS", "CRAFT", "ANN", "web"]
favorite_subjects.append("MATHS(i am lying)")
favorite_subjects.insert(5,"OS")
favorite_subjects.sort()
print(favorite_subjects, len(favorite_subjects))

print("Task 3 ============================================================================")
#a) Create a tuple named weekdays containing the names of five weekdays.
#b) Attempt to change the second element of the tuple and observe/print the error message Python gives (add this as a comment).
#c) Create a dictionary named subject_credits that maps at least four subject names to their credit hours (e.g., "AIC380": 3).
#d) Update the credit hours of one existing subject and add one new subject-credit pair to the dictionary. Print the final dictionary.

weekdays = ("moday", "tuesday","wednesday","thursday","friday")
print(weekdays)

subject_credits={"OS":3, "ML":3, "CAL":3, "LA":2}
subject_credits["OS"]=4
subject_credits["AI"]=3
print(subject_credits)


print("Task 4 ============================================================================")
#a) Create a list of the numbers 1 to 15 using range() and list().
#b) Extract and print the first 4 elements and the last 4 elements using slicing.
#c) Extract and print every 3rd element of the list starting from index 0.
#d) Print the list in reverse order using slicing (without using the reverse() function).

num = list(range(1, 16))
print("First 4 elements:", num[:4])
print("Last 4 elements:", num[-4:])
print("Every 3rd element:", num[::3])
print("List in reverse order:", num[::-1])

print("Task 5 ============================================================================")
#a) Store your full name in a string variable and use split() to separate it into first_name and last_name.
#b) Use an f-string to print a formatted sentence: "My first name is <first_name> and my last name is <last_name>."
#c) Write a loop that counts and prints the number of vowels (a, e, i, o, u) in your full name.
#d) Convert your full name to uppercase and lowercase separately, and check (using ==) whether the uppercase version equals the lowercase version.

name = "Shumaim Iftikhar"
first_name, last_name = name.split()
print(f"My first name is {first_name} and my last name is {last_name}.")

for char in name:
  if char in "aeiouAEIOU":
    print(char)

uppercase_name = name.upper()
lowercase_name = name.lower()
print(uppercase_name == lowercase_name)


print("Task 6 ============================================================================")
#a) Take a marks value (0–100) stored in a variable and use if / elif / else to print the grade (A, B, C, D, or F) based on standard grading criteria.
#b) Write a condition to check whether a given number is positive, negative, or zero.
#c) Write nested if statements to check whether a given year is a leap year.
#d) Combine two conditions using the and / or operators to check whether a student is eligible for a scholarship (e.g., CGPA > 3.5 and attendance > 80%).

mark=67
if mark >= 90:
  print("A")
elif mark >= 80:
  print("B")
elif mark >= 70:
  print("C")
elif mark >= 60:
  print("D")

if mark > 0:
  print("positive")
elif mark < 0:
  print("negative")
else:
  print("zero")

year=2027

if year % 4 == 0:
  if year % 100 == 0:
    if year % 400 == 0:
      print("leap year")
    else:
      print("not a leap year")
  else:
    print("leap year")
else:
  print("not a leap year")

cgpa=3.32
attendence=83
if cgpa > 3.5 and attendance > 80:
  print("eligible for scholarship")
else:
  print("not eligible for scholarship")


print("Task 7 ============================================================================")
#a) Use a for loop to print all numbers from 1 to 30, but skip every multiple of 4 using continue.
#b) Use a for loop to print numbers from 1 to 50, and stop the loop as soon as a number greater than 20 is found, using break.
#c) Use a while loop to print the multiplication table (1 to 10) of a number provided in a variable.
#d) Write a loop that calculates and prints the sum of all even numbers between 1 and 100.

for i in range(1, 31):
  if i % 4 == 0:
    continue
  print(i)

for i in range(1, 51):
  if i > 20:
    break
  print(i)

while True:
  num = int(input("Enter a number: "))
  for i in range(1, 11):
    print(num, "x", i, "=", num * i)

  choice = input("Do you want to continue? (y/n): ")
  if choice.lower() != "y":
    break

sum = 0
for i in range(2, 101, 2):
    sum += i
    print(sum)

print("Task 8 ============================================================================")
#a) Write a function calculate_area(length, width) that returns the area of a rectangle, with width having a default value of 1.
#b) Call the function three times: once with both arguments, once with only length, and once storing the result in a variable and printing it.
#c) Write a function is_even(number) that returns True if the number is even and False otherwise.
#d) Write a function that takes a list as a parameter, adds one item to it inside the function, and prints the list before and after the function call to demonstrate how lists are passed by reference.

def calculate_area(length, width=1):
  return length * width

print(calculate_area(5, 10))
print(calculate_area(5))
result = calculate_area(5, 10)
print(result)

def is_even(number):
  if number % 2 == 0:
    return True
  else:
    return False

print(is_even(10))
print(is_even(11))

def add_item(lst):
  lst.append("new item")
  print("List after function call:", lst)

my_list = [1, 2, 3]
print("List before function call:", my_list)
add_item(my_list)


print("Task 9 ============================================================================")
#a) Create a class Course with an __init__ method that stores course_name and credit_hours, and a method display_info() that prints these details.
#b) Create two objects of the Course class with different data and call display_info() on both.
#c) Create a class LabCourse that inherits from Course and adds an extra attribute lab_hours, using super().__init__() to reuse the parent constructor.
#d) Override the display_info() method in LabCourse so that it prints all the parent's details plus the lab_hours, and demonstrate it by creating one LabCourse object.

class Course:
  def __init__(self, course_name, credit_hours):
    self.course_name = course_name
    self.credit_hours = credit_hours

  def display_info(self):
    print("Course Name:", self.course_name)
    print("Credit Hours:", self.credit_hours)

course1 = Course("AI", 3)
course2 = Course("OS", 3)
course1.display_info()
course2.display_info()

class LabCourse(Course):
  def __init__(self, course_name, credit_hours, lab_hours):
    super().__init__(course_name, credit_hours)
    self.lab_hours = lab_hours
    def display_info(self):
      super().display_info()
      print("Lab Hours:", self.lab_hours)

lab_course = LabCourse("AI", 3, 2)
lab_course.display_info()
```
